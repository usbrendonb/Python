```python
# matplotlib
import matplotlib.pyplot as plt
```

```python
# LineChart

months = ["Jan","Feb","Mar","Apr","May"]
sales =[25000,28000,35000,42000,47000]

plt.plot(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
# BarChart

departments = ["HR","IT","Finance","Sales"]
employees = [12,30,18,20]

plt.bar(departments, employees)
plt.title("Employees by department")
plt.xlabel("Departments")
plt.ylabel("Sales")
plt.show()
```

```python
# PieChart

# Scenario: Market Share

brands = ["Apple","Samsung","OnePlus","Others"]
share = [40,30,20,10]

plt.pie(share,
        labels=brands,
        colors=["red","green","blue","yellow"],
        autopct="%1.1f%%")
plt.title("Market Share")
plt.xlabel("Brands")
plt.ylabel("Share")
plt.show()
```

```python
# Histogram Scenario

salary = [35000,42000,50000,45000,62000,58000,47000,39000,51000,65000,54000]

plt.hist(salary,bins=8)
plt.title("Salary Distribution")
plt.show()
```

```python
# Customizing Charts
plt.plot(months,sales,marker='o')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv("/content/sample_data/Smart_Retail_Sales_Dataset - Smart_Retail_Sales_Dataset.csv")

month_order = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

df = pd.DataFrame(data)
monthly_sales = (df.groupby("Month")["Sales_Amount"].sum()
.reindex(month_order)
.dropna()
)

print("Monthly Sales line chart")
plt.plot(monthly_sales.index,monthly_sales.values,marker='o')
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.show()


print("Category sales Bar Chart")
category_sales = df.groupby("Category")["Sales_Amount"].sum()

plt.bar(category_sales.index, category_sales.values)

plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.grid(axis="y")

plt.show()

print("Product Contribution Pie Chart")

product_sales = df.groupby("Product")["Sales_Amount"].sum()

plt.pie(
    product_sales.values,
    labels=product_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Contribution")

plt.show()

print("Sales Distribution Histogram")

plt.hist(df["Sales_Amount"],bins=10)
plt.title("Sales Distribution")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")

```

Mini Project: Hospital Patient Analytics Dashboard

```python

import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv("/content/sample_data/Dataset - Sheet1.csv")

df = pd.DataFrame(data)

print("Printing Dataset: ")
print()
print(df)
print()

print("Printing Dataset first five lines: ")
print()
print(df.head(5))
print()

print("Printing Dataset info: ")
print()
print(df.info())
print()

print("Printing Dataset Description: ")
print()
print(df.describe())
print()

print("Printing Total Treatment Cost: ")
print()
print(df['Treatment_Cost'].sum())
print()

print("Printing Average Treatment Cost: ")
print()
print(df['Treatment_Cost'].mean())
print()

print("Printing highest Treatment Cost: ")
print()
print(df['Treatment_Cost'].max())
print()

print("Printing lowest Treatment Cost: ")
print()
print(df['Treatment_Cost'].min())
print()

print("Printing Average patient Age: ")
print()
print(df['Age'].mean())
print()

print("Number of patients admitted for more than 7 days: ")
print()
over_7 = np.sum(df["Days_Admitted"]  >7)
print(over_7)
print()

print("Printing Department-wise Treatment Cost: ")
print()
print(df.groupby("Department")["Treatment_Cost"].sum())
print()

print("Printing Month-wise Patient Count: ")
print()
print(df.groupby("Month")["Patient_ID"].count())
print()

print("Printing City-wise Patient Count: ")
print()
print(df.groupby("City")["Patient_ID"].count())
print()

#Monthly Patient Count (Line Chart)
print("Monthly Patient Count (Line Chart)")
monthly_patients = df.groupby("Month").size()
plt.plot(monthly_patients.index, monthly_patients.values, marker="o")
plt.title("Monthly Patient Count")
plt.xlabel("Months")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()

#Department-wise Treatment Cost (Bar Chart)
print("Department-wise Treatment Cost (Bar Chart)")
department_cost = df.groupby("Department")["Treatment_Cost"].sum()
plt.bar(department_cost.index, department_cost.values)
plt.title("Department-wise Treatment Cost")
plt.xlabel("Department")
plt.ylabel("Total Treatment Cost")
plt.grid(True)
plt.show()

print("Department-wise Patient Distribution (Pie Chart)")
department_counts = df["Department"].value_counts()
plt.pie(
    department_counts.values,
    labels=department_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Department-wise Patient Distribution")
plt.show()

print("Treatment Cost Distribution (Histogram)")
plt.hist(df["Treatment_Cost"], bins=10)
plt.title("Treatment Cost Distribution")
plt.xlabel("Treatment Cost")
plt.ylabel("Frequency")
plt.show()
```
