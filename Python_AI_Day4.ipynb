```python

import pandas as pd

data = {
    "Name": ["Rajul","Priya","Aman","Sneha","Rohan"],
    "Department": ["IT","HR","Finance","IT","HR"],
    "Salary": [60000,45000,70000,65000,50000]
}

df = pd.DataFrame(data)
print(df)
print()
print("First Three:")
print(df.head(3))
print()
print("Average Salary:")
print(df['Salary'].mean())
print()
print("Highest Paid Employee")
print(df.loc[df['Salary'].idxmax()])
print()
print("All IT employees")
print(df[df['Department'] == 'IT'])
print()
print("Bonus")
df["Bonus"] = df["Salary"] * 0.10
print(df)
```

```python
from ast import Index
import pandas as pd

data = {
    "Name": ["Aisha","Rahul","Priya","Aman","Sneha"],
    "Math": [85,72,95,68,89],
    "Science": [90,80,91,74,86],
    "English":[88,75,94,70,90]
}

df = pd.DataFrame(data)
print('Averages:')
average= df['Average'] = df[['Math', 'Science', 'English']].mean(axis=1).round(0)
print(df.to_string(index=False))

print()
print('Top Students per subject:')
top_students = df.set_index('Name')[['Math', 'Science', 'English']].idxmax()
print(top_students)

print()
print('Students averaging above 85:')
above_85 = df[df['Average'] > 85]
print(above_85)

print()
print("Adding letter grade column with grade")
add_grade = pd.cut(
    df["Average"],
    bins=[0, 65, 75, 85, 95, 101],
    labels=["F", "D", "C", "B", "A"],
    include_lowest=True,
    right=False
)

df["Grade"] = add_grade.astype(str)
print(df.to_string())

print()
print("Final Report")
sort_average = df.sort_values(by="Average", ascending=False)

print(sort_average)
```
