```python
#  Hospital Appointment Assistant

from datetime import datetime


def hospital_chatbot():

    doctors = {
        "Dr. One": "Monday - Friday (9:00 AM - 1:00 PM)",
        "Dr. Two": "Monday - Saturday (10:00 AM - 4:00 PM)",
        "Dr. Three": "Tuesday - Sunday (9:00 AM - 3:00 PM)",
        "Dr. Four": "Monday - Friday (1:00 PM - 6:00 PM)"
    }


    specialty_to_doctor = {
        "Cardiology": "Dr. One",
        "Dermatology": "Dr. Two",
        "Pediatrics": "Dr. Three",
        "Orthopedics": "Dr. Four"
    }


    print("=" * 50)
    print("      Hospital Appointment Assistant")
    print("=" * 50)

    while True:

        print("\nChoose an option:")
        print("1. Book Appointment")
        print("2. View Doctor Availability")
        print("3. Hospital Timings")
        print("4. Frequently Asked Questions")
        print("5. Symptom Checker")
        print("6. Exit")

        choice = input("\nEnter your choice (1-6): ")


        if choice == "1":

            print("\n--- Book Appointment ---")

            patient = input("Enter Patient Name: ")

            print("\nAvailable Doctors:")
            for doctor in doctors:
                specialty = next(
                  spec for spec, doc in specialty_to_doctor.items()
                  if doc == doctor
              )
                print(f"- {doctor}: {specialty}")

            while True:
                date = input("Enter Appointment Date (DD/MM/YYYY): ")

                try:
                    datetime.strptime(date, "%d/%m/%Y")
                    break
                except ValueError:
                    print("Invalid date. Please use DD/MM/YYYY.")


            while True:
                time = input("Enter Appointment Time (HH:MM AM/PM): ")

                try:
                    datetime.strptime(time, "%I:%M %p")
                    break
                except ValueError:
                    print("Invalid time. Please use HH:MM AM/PM.")

            print("\nAppointment Booked Successfully!")
            print("--------------------------------")
            print(f"Patient : {patient}")
            print(f"Doctor  : {doctor}")
            print(f"Date    : {date}")
            print(f"Time    : {time}")


        elif choice == "2":

            print("\n--- Doctor Availability ---")

            for doctor, schedule in doctors.items():
                print(f"{doctor}: {schedule}")

        elif choice == "3":

            print("\n--- Hospital Timings ---")
            print("Monday - Sunday")
            print("8:00 AM - 8:00 PM")

        elif choice == "4":

            print("\n--- Frequently Asked Questions ---")

            print("\nQ: Do I need an appointment?")
            print("A: Walk-ins are accepted, but appointments are recommended.")

            print("\nQ: What documents should I bring?")
            print("A: Bring your ID, insurance card, and previous medical records.")

            print("\nQ: Do you accept insurance?")
            print("A: Yes, we accept most major insurance providers.")

            print("\nQ: What is the emergency number?")
            print("A: Call 911 for emergencies.")


        elif choice == "5":

            print("\n--- Symptom Checker ---")

            symptom = input("Enter your main symptom: ").strip().lower()

            if symptom in ["chest pain", "heart pain", "palpitations"]:
                department = "Cardiology"

            elif symptom in ["skin rash", "acne", "itching"]:
                department = "Dermatology"

            elif symptom in ["fever", "cough", "child fever"]:
                department = "Pediatrics"

            elif symptom in ["bone pain", "joint pain", "fracture"]:
                department = "Orthopedics"

            else:
                department = "General Medicine"

            print(f"\nSuggested Department: {department}")

            if department in specialty_to_doctor:
                doctor = specialty_to_doctor[department]
                print(f"Recommended Doctor: {doctor}")
                print(f"Availability: {doctors[doctor]}")
            else:
                print("Please visit the General Medicine department.")


        elif choice == "6":

            print("\nThank you for using the Hospital Appointment Assistant.")
            print("Have a great day!")
            break
        else:
            print("\nInvalid choice. Please enter a number between 1 and 6.")


# Start the chatbot
hospital_chatbot()
```

```python
# Customer Purchase Prediction

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.metrics import (accuracy_score,confusion_matrix,classification_report)

#question = input("Do you need to upload a file? Yes/No")#
#if question == "yes":
#  uploaded = files.upload()
#else:
#  print("No file uploaded")
#  exit()


df = pd.read_csv("Customer_Purchase_Prediction.csv")

print("First 10 Rows")
print(df.head(10))

print("\nDataset Shape")
print(df.shape)

print("\nData Types")
print(df.dtypes)

print("\nDataset Information")
print(df.info())

print("\nSummary Statistics")
print(df.describe())

print("\nMissing Values")
print(df.isnull().sum())

df["Purchased"] = df["Purchased"].map({
    "No": 0,
    "Yes": 1
})


X = df.drop(columns=["Customer_ID", "Purchased"])
y = df["Purchased"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("\nAccuracy")
print(accuracy_score(y_test, y_pred))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))


comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print("\nActual vs Predicted")
print(comparison.head(10))


new_customer = pd.DataFrame({
    "Age": [30],
    "Monthly_Income": [6500],
    "Website_Visits": [12],
    "Time_Spent_Min": [45],
    "Previous_Purchases": [4]
})

prediction = model.predict(new_customer)

if prediction[0] == 1:
    print("\nPrediction: Customer WILL PURCHASE.")
else:
    print("\nPrediction: Customer WILL NOT purchase.")
```

```python
# House Price Prediction using Linear Regression

import pandas as pd
from google.colab import files
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

question = input("Do you need to upload a file? Yes/No")
if question == "yes":
 uploaded = files.upload()
else:
 print("No file uploaded")
 exit()

df = pd.read_csv("House_Price_Prediction.csv")

print("First 10 Rows")
print(df.head(10))

print("\nDataset Shape")
print(df.shape)

print("\nData Types")
print(df.dtypes)

print("\nDataset Information")
print(df.info())

print("\nSummary Statistics")
print(df.describe())

print("\nMissing Values")
print(df.isnull().sum())

X = df[[
    "Area_sqft",
    "Bedrooms",
    "Bathrooms",
    "Age_Years",
    "Distance_City_km"
]]

y = df["Price_Lakhs"]
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\nModel Evaluation")
print("---------------------------")
print("Mean Absolute Error :", round(mae, 2))
print("Mean Squared Error  :", round(mse, 2))
print("R² Score            :", round(r2, 4))

comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred.round(2)
})

print("\nActual vs Predicted")
print(comparison.head(10))


new_house = pd.DataFrame({
    "Area_sqft": [2200],
    "Bedrooms": [4],
    "Bathrooms": [3],
    "Age_Years": [8],
    "Distance_City_km": [6]
})

predicted_price = model.predict(new_house)

print("\nPredicted House Price")
print("$", round(predicted_price[0], 2))
```
