```python
#AI Example

keywords=["buy","price","discount"]

sentence=input("Enter text : ")

for word in keywords:
  if word in sentence.lower():
    print("Marketing Intent Detected")
    break
```

```python
education=["course","trainint","collete"]
shopping=["buy","price","discount"]
text=input("Enter Website description")

if any(word in text.lower() for word in education):
  print("Education Website")
elif any(word in text.lower() for word in shopping):
  print("Shoping Website")
else:
  print("General website")
```
