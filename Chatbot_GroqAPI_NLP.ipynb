```python
# Chatbot GroqAPI NLP
# Project 2 – AI Chatbot
# Using the Groq API:
# Students will:
# Import the Groq library
# Load the API key
# Send a user question
# Display the AI-generated response
```

```python
!pip install groq
from groq import Groq
from google.colab import userdata
api_key = userdata.get("GROQ_API")
userdata.get('GROQ_API')
print("API Key Loaded Successfully")
```

```python

client = Groq(
    api_key=userdata.get("GROQ_API")
)

topic=input("Enter Topic : ")
audience=input("Enter Audience (Beginner, Advanced etc...) : ")
output_format=input("Enter format for the output (Bullet, Table, Quiz etc...) : ")

prompt=f"""
Explain {topic}
Audience {audience}
Format {output_format}
"""

response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role":"user",
         "content":prompt
        }
    ]
)
print(response.choices[0].message.content)
```
